## Ejercicio 3.9 — Comparación Matilda 1 (2019) vs. Matilda 3 (2021)
Compare los 15 unigramas y bigramas más frecuentes de cada libro, primero completos y luego excluyendo palabras funcionales (defina su lista de *stopwords* y justifíquela). Responda con evidencia:
1. ¿Por qué los unigramas más frecuentes son casi idénticos? (Zipf, palabras funcionales.)
2. ¿Qué bigramas de contenido cambian entre 2019 y 2021 y qué reflejan?
3. ¿Qué porcentaje de los tipos de bigrama de Matilda 3 **no** aparece en Matilda 1? ¿Qué implica para un modelo MLE?

````python
# Su código aquí
STOP = set("""de la el en y a los las del que se por un una con para al es su lo como más o
sus le ha son este esta entre sobre también ser pero sin muy así desde otros otras
cada hay fue han ya nos mi mis me te ni sí si no e u era ese todo""".split())

def sin_stop_uni(m, k):
    return [(w, c) for w, c in m.c1.most_common() if w not in (BOS, EOS) and w not in STOP][:k]

def sin_stop_bi(m, k):
    return [(b, c) for b, c in m.c2.most_common()
            if BOS not in b and EOS not in b and b[0] not in STOP and b[1] not in STOP][:k]

M = {i: ModeloNGrama().entrenar(libro[i]) for i in (1, 3)}
for i in (1, 3):
    print(f"--- Libro {i}: unigramas        :", M[i].top_unigramas(15))
    print(f"--- Libro {i}: bigramas         :", M[i].top_bigramas(15))
    print(f"--- Libro {i}: unigramas s/stop :", sin_stop_uni(M[i], 15))
    print(f"--- Libro {i}: bigramas s/stop  :", sin_stop_bi(M[i], 15))

b1 = {b for b in M[1].c2 if BOS not in b and EOS not in b}
b3 = {b for b in M[3].c2 if BOS not in b and EOS not in b}
print(f"Tipos de bigrama de Matilda 3: {len(b3)}; no están en Matilda 1: {len(b3-b1)} = {100*len(b3-b1)/len(b3):.1f}%")
````

>  **Análisis de la pareja:**
por qué de, la, que, y dominan (palabras funcionales, ley de Zipf), qué bigramas de contenido cambian de 2019 a 2021 (por ejemplo, estados unidos o américa latina en el libro 1, y universidad nacional o mis padres en el 3), y qué significa que el 76.7 % de los bigramas sean nuevos para un MLE.

Entrene con **Matilda 1 + 2** y evalúe en **Matilda 3**: (a) sin suavizado, (b) con add-1. Calcule también la perplejidad add-1 sobre el propio entrenamiento. Explique: ¿por qué la MLE da ∞? ¿Por qué add-1 da un valor tan alto? ¿Qué técnica del capítulo lo reduciría?
